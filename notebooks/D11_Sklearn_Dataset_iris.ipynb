# 붓꽃 데이터 요약 분석(Summary)

```
import pandas as pd
from sklearn.datasets import load_iris
```

```
iris = load_iris()
# type(iris)
```

```
# iris.data, iris['target']
```

```
iris['data'].shape, iris.target.shape
```

```
print(iris.keys())
print(iris['feature_names'])
print(iris['target_names'])
print(iris['filename'])
# print(iris['DESCR'])
```

```
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
iris_df['Target'] = iris.target
# iris_df
# iris_df.iloc[[0, 1, 2, 50, 51, 52, 100, 101, -1]]
iris_df.iloc[[0, 1, 2, 47, 48, 49, 50, 51, 52, 97, 98, 99, 100, 101, 102, 147, 148, -1]]
```

```
iris_df.info()
```

```
iris_df.describe()
```

```
round(iris_df.groupby('Target').describe(), 2) # .T
```

```
col_dict = {0:'blue', 1:'green', 2:'red'}
cols = [col_dict[x] for x in iris_df.Target]
pd.plotting.scatter_matrix(iris_df, color=cols, figsize=(12,12), marker='o') ;
```

```
import seaborn as sns
import matplotlib.pyplot as plt

sns.pairplot(hue='Target', data=iris_df, palette='dark', markers=["o", "D", "X"]) #, height=2, corner=True)
# plt.gcf().set_size_inches(18, 17)
plt.show()
```
