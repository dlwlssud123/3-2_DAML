## **조건문(if, if ~ else, if ~ elif ~ else)**

## **함수**

```python
def add_times(x, y=1) :     # 매개변수에 default 값을 줄 수 있음
    return x+y, x*y         # return은 함수 결과를 반환, 여러 개 반환 가능 (tuple로 반환됨)

a, b = add_times(6, 7)      # add_times(6)의 경우 y는 자동으로 1로 주어짐
print("x+y={} \nx*y={}".format(a, b))
```

```python
def multi_args(*x) :        # 매개변수의 수가 가변적일 때 *매개변수명 사용
    sum = 0
    for i in x :
        sum = sum + i
    print("입력된 변수는 {}이며, 합은 {}이다".format(x, sum))
multi_args(1,2, 3)
multi_args(2, 3, 4, 5, 6, 7, 8, 9)
```

- **람다(lambda) 함수**

```python
def return_func(t) :
    if t == 0 :
        return lambda x, y : x+y     # 매개변수가 두 개인 함수 반환
    elif t == 1 :
        return lambda x, y : x*y

f = return_func(0)        # 변수 f는 매개변수 두개인 함수를 값으로 가짐
print(f(2, 7))
f = return_func(1)
print(f(2, 7))

lf = lambda x, y: x**2 +3*y
def func(f, a, b) :
    print(f(a, b))
func(lf, 3, 2)
```

## **클래스와 객체**
객체지향 프로그래밍
- 세상에 존재하는 여러 종류들에 속하는 객체들은 종류에 따라 서로 다른 속성과 행동 양식이 존재
- 같은 종류에 속하는 객체들은 모두 같은 송성과 행동 양식을 가지지만, 서로 다른 속성값을 가지고 있어서 서로 구별됨
- 특정 객체를 생성/조작을 위해
    - 특정 종류에 속한 모든 객체들의 공통된 성질과 동작을 정의
    - 특정 객체의 개별적 속성값을 부여
    - 객체 단위로 속성과 동작을 다루기

```python
class Rectangle:
    def __init__(self, w, h):
        self.W = w
        self.H = h
    def area(self):
        return self.W * self.H
    def perimeter(self):
        return (self.W + self.H)*2

r1 = Rectangle(10, 20)
a = r1.area()
p = r1.perimeter()
print("r1의 면적은 {}, 둘레는 {}".format(a, p))

r2 = Rectangle(8, 22)
print("r2의 면적은 {}, 둘레는 {}".format(r2.area(), r2.perimeter()))

print(type(r1))
```

```python
class Calc:
    def __init__(self, arg1=0, arg2=0):
        self.arg1 = arg1
        self.arg2 = arg2
    def ins_data(self, arg1, arg2):
        self.arg1 = arg1
        self.arg2 = arg2
    def add(self):
        return self.arg1+self.arg2
    def mul(self):
        return self.arg1*self.arg2
    def sub(self):
        return self.arg1-self.arg2
    def div(self):
        if (self.arg2 == 0) :
            return "Zero Division Error"
        else :
            return self.arg1/self.arg2

a = Calc()
a.ins_data(3, 4)
a.add(), a.div()
```

- **속성 상속(Property Inheritance)**
- **메소드 오버라이딩(Method Overriding)**

```python
class CalcExp(Calc) :   # CalcExp는 Calc의 서브클래스(자식 클래스)
    def pow(self):
        return self.arg1**self.arg2
    def div(self):      # 메소드 오버라이딩
        if (self.arg2 == 0) :
            return "0으로 나누면 안돼!"
        else :
            return self.arg1/self.arg2

c = CalcExp()
c.ins_data(2,4)
c.mul(), c.pow(), c.div()
```

```python
b=Calc()
d=CalcExp()
b.div(), d.div()
```

- **연산자 오버로딩(Operator Overloading)**

```python
class coordinate:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def __add__(self, other):     # 파이썬 magic method, 연산자 overloading
        return coordinate(self.x + other.x, self.y + other.y)
    def __str__(self):
        return '<{}, {}>'.format(self.x, self.y)

coord1 = coordinate(5, 7)
coord2 = coordinate(3, 9)
coord3 = coord1 + coord2        #  coord3 = cord1.__add__(coord2)
print(coord1)
print(coord2)
print(coord3)
```
