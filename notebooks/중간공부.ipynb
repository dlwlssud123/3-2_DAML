# 함수
- 같은 일 여러 번 수행
- def 절로 실행
## *args
전달하는 인수의 개수가 몇 개인지 미리 알 수 없을 때 사용, 전달된 값을은 함수 내부에서 튜플로 묶임
## **kwargs
키=값 형태로 여러 개의 매개변수를 전달받을 대 싸용, 함수 내부에서 딕셔너리로 취급
## lambda 함수



```python
def add_many(*args):
    result = 0
    for i in args:
        result += i
    return result

def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f'{key}: {value}')

print(add_many(1, 2, 3, 4, 5))
print(add_many(10, 20, 30))

dict1 = {
    'name': '홍길동',
    'age': 20,
    '대학': '서울대학교'
}

print_info(**dict1)
print_info(a = 1, b = 2, c = 3)

add_lambda = lambda x, y: x+y
print(add_lambda(10, 20))

# 정렬할 때 기준 정하기
students = [('철수', 20),
            ('영희', 19),
            ('민수', 21)]
sorted_age = sorted(students, key = lambda x: x[1])

words = ['apple', 'banana', 'cherry', 'date']
sorted_words = sorted(words, key = lambda w: len(w))

print(sorted_age)
print(sorted_words)

# 리스트 값 일괄 변화
numbers = [1, 2, 3, 4, 5]
# map(func, iterable): iterable의 각 요소에 func를 적용한 결과를 map 객체로 반환
squared_numbers = list(map(lambda x: x**2, numbers))
print(squared_numbers)

price = [100, 200, 300, 400, 500]
formaated = list(map(lambda x: f"{x}원", price))
print(formaated)

# 조건에 맞는 값만 
numbers = list(range(0, 100, 5))
even_nums = list(filter(lambda x: x % 2 == 0, numbers))
print(even_nums)

# 삼항 연산자
check_even = lambda x: '짝수' if x % 2 == 0 else '홀수'

print(check_even(10))
```

# Python object oriented programming
- 클래스 
- 클래스 상속
- 오버라이딩
- 연산자 오버로딩

```python
class Calc:
    
    def __init__(self, a = 0, b = 0):
        self.a = a
        self.b = b
    
    def add(self):
        return self.a + self.b 
    
    def div(self):
        if self.b == 0:
            return 'Zero Division Error'
        else:
            return self.a / self.b

class CalcExp(Calc): # Calc 상속
    def pow(self):
        return self.a ** self.b
    
    def div(self): # 부모의 메소드 오버라이딩
        if self.b == 0:
            return '닥치'
        else:
            return self.a / self.b
        
c1 = Calc(10, 0)
c2 = CalcExp(10, 0)
print(c1.div()) 
print(c2.div())

# 매직 메소드와 연산자 오버로딩
class Coordinate:
    
    def __init__(self, x, y):
        self.x = x
        self.y = y
        
    def __add__(self, other): # '+' 연산자 오버로딩
        return Coordinate(self.x + other.x, self.y + other.y)
    
    def __str__(self): # print() 호출 시 출력 형태 지정
        return f'<{self.x}, {self.y}>'
    
p1 = Coordinate(10, 20)
p2 = Coordinate(30, 40)
p3 = p1 + p2
print(p3)
```

```python
import pandas as pd

# 1. 기본 인적사항 테이블
df_info = pd.DataFrame({
    '이름': ['김', '이', '박', '최'],
    '성별': ['남', '여', '남', '여'],
    '주소': ['대구', '대구', '경산', '포항']
})

# 2. 시험 점수 테이블 (새로운 학생 '정' 포함, '최'는 누락)
df_score = pd.DataFrame({
    '이름': ['김', '이', '박', '정'],
    '중간점수': [85, 82, 90, 70],
    '기말점수': [88, 89, 98, 75]
})

# pd.merge()
# on에 키, how에 조인 방식
merged = pd.merge(df_info, df_score, on = '이름', how = 'inner')
display(merged)
# left/right/outer

merged_left = pd.merge(df_info, df_score, on = '이름', how = 'left')
display(merged_left)

merged_right = pd.merge(df_info, df_score, on = '이름', how = 'right')
display(merged_right)

merge_outer = pd.merge(df_info, df_score, on = '이름', how = 'outer')
display(merge_outer)
```

```python
df_dept = pd.DataFrame({
    '학생명': ['김', '이', '박'],
    '학과': ['컴공', '인공지능', '소프트웨어']
})

merged_diff_key = pd.merge(df_info, df_dept, left_on='이름', right_on='학생명')
# 병합 후 중복되는 키 컬럼 제거
merged_diff_key.drop('학생명', axis=1, inplace=True)
```

```python
# pd.concat
# 위아래로 행 붙이기 

df_new_students = pd.DataFrame({
    '이름': ['강', '조'],
    '성별': ['남', '여'],
    '주소': ['서울', '부산']
})

concat_rows = pd.concat([df_info, df_new_students], axis = 0, ignore_index = True)
display(concat_rows)

# 옆으로 붙이기
df_extra = pd.DataFrame({
    '토익점수': [850, 920, 780, 890],
    '학점': [3.8, 4.2, 3.5, 4.0]
})
concat_cols = pd.concat([df_info, df_extra], axis = 1)
display(concat_cols)
```

```python
import pandas as pd

df = pd.DataFrame(
    {
        "총점": [88, 89, 98, 77],
        "성별": ["남", "여", "남", "여"],
        "나이": [19, 21, 20, 21],
        "주소": ["대구", "대구", "경산", "포항"],
        "키": [175.0, 160.0, 180.0, 166.5],
    },
    index=["김", "이", "박", "최"],
)

# 1) 메타 데이터 요약 및 통계 함수
df.info()
display(df.describe())

# 축 이름 지정
df.index.name = '학생명'
df.columns.name = '특성'
display(df)

# 전치 행렬
display(df.T)

# 문자열 전용 메서드
dae_students = df[df['주소'].str.contains('대')]
print(dae_students[['성별', '주소']])

```

```python
# 시각화 기초
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(1, 10)
y = x**2

# 선 그래프
plt.plot(x, y, color = 'blue', linestyle = '--', marker = 'o', label = 'Trend')
plt.title('Line plot')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.show()
```

```python
plt.scatter(x, y, color = 'red', marker = 'x', label = 'Scatter')
plt.title('Scatter plot')
plt.show()
```

```python
# 막대 그래프
categories = ['A', 'B', 'C', 'D', 'E']
plt.bar(categories, [5, 7, 3, 8, 6], color = 'green', label = 'Bar')
plt.title('Bar plot')
plt.show()
```
