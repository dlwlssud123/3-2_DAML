## 시리즈(Series)
- 일련의 나열된 값들 : 하나의 열(column)을 구성

```python
import numpy as np
import pandas as pd

s = pd.Series([88, 39, 79, 88, 27])
print(s)
s
```

```python
dic_data = {"김": 88, "이": 89, "박":80, "최":98, "정": 75}
s1 = pd.Series(dic_data)
s1
```

```python
print(s1['이'])
print(s1[2])
print(s1[['김', '정']])
print(s1[2:4])
```

- **Pandas에서 관계/논리 연산**
    - ==, !=, <, >, <=, >=
    - and, or, not은 **&, |, ~을 사용함**

```python
s1[(s1 >= 80) & (s1 < 90)]
```

```python
s1[~(s1 >= 80) |  ~(s1 < 90)]
```

```python
l_data = [88, 89, 98, 77]
s2 = pd.Series(l_data)
print(s2, '\n')
print(s2[1:3], '\n')
print(s2.values, '\n')
print(s2.index, '\n')

s3 = pd.Series(l_data, index=['Kim', 'Lee', 'Park', 'Choi'])
# s3.index.name = '이름'
print(s3, '\n')
print(s3.index, '\n')
print(s3['Lee':'Choi'], '\n')   # 인덱스 이름으로 범위를 주면 마지막 인덱스 포함됨
```

### **데이터 프레임의 조건 필터링(조건 검색/추출)**

- df[ ], df.loc[ ], df.iloc[ ]를 선택적으로 활용

        - df[ ]은 행, 열을 동시 지정 불가
        - df.loc[ ], df.iloc[ ]은 행렬 동시 지정 가능


```python
import pandas as pd
df = pd.DataFrame({"총점": [88, 89, 98, 77],
                   "성별": ["남", "여", "남", "여"],
                   "나이": [19, 21, 20, 21],
                   "주소": ["대구", "대구", "경산", "대전"],
                   "키": [175., 160., 180., 166.5]},
                  index = ['김', '이', '박', '최']
                )
df
```

```python
df['이':'최']
```

```python
df[df['총점'] <= 90] [['성별', '나이', '주소']]  # 주석을 지우면 열 스마트인덱싱
```

- **다중 조건**
    - 논리연산 : &, |, ~
    - 관계연산 : ==, !=, <, >=, isin(), ...
    - ***논리연산은 반드시 괄호로 분리***

```python
df[(df['총점'] >= 80) & (df['키'] >= 170)] [['성별', '나이', '주소']]  # 주석을 지우면 열 스마트인덱싱
```

위의 조건추출에서 조건부분(필터)만 실행시켜보면 부울 값의 시리즈(***부울 인덱싱***)를 출력함

```python
filter = (df['총점'] >= 80) & (df['키'] >= 170)
print(filter)

df[filter]  # 데이터프레임에서 filter 행이 True인 부분만 출력됨
```

```python
df[~filter]
```

## **피봇(Pivot) 테이블**
- 데이터 열 중에서 두 개의 열을 각각 행 인덱스, 열 인덱스로 사용하여 데이터를 조회하여 펼쳐놓은 것
- pivot 메소드:
    - 1st 인수로는 행 인덱스로 사용할 열 이름,
    - 2nd 인수로는 열 인덱스로 사용할 열 이름,
    - 3rd 인수는 데이터로 사용할 열 이름

```python
df = pd.DataFrame({"총점": [88, 89, 98, 77],
                   "성별": ["남", "여", "남", "여"],
                   "나이": [19, 21, 20, 21],
                   "주소": ["대구", "대구", "경산", "포항"],
                   "키": [175., 160., 180., 166.5]},
                  index = ['김', '이', '박', '최']
                )
df
```

```python
df.pivot(index='주소', columns="성별", values='총점')
```

```python
df.pivot(index=['주소', '성별'], columns="나이", values=['총점', '키']).fillna('~')
```

## **그룹 분석**
- groupby 메소드 : 인수는
    - 열 or 열 리스트
    - 행 인덱스

- 연산결과로 그룹 데이터를 나타내는 ***GroupBy 클래스 객체***를 반환
    - 이 객체에는 그룹별로 연산을 할 수 있는 그룹연산 메서드가 있음
    - groupby 결과, 즉 GroupBy 클래스 객체의 뒤에 붙일 수 있는 그룹연산 메서드는 다양함
        - size, count, mean, median, min, max, sum, prod, std, var, quantile, first, last 등
        - 이 외에도 agg(aggregate), apply, transform 등이 많이 사용됨

```python
# Example
df2 = pd.DataFrame({
    'key1': ['A', 'A', 'B', 'B', 'A'],
    'key2': ['one', 'two', 'one', 'two', 'one'],
    'data1': [1, 2, 3, 4, 5],
    'data2': [10, 20, 30, 40, 50]
})
df2
```

```python
gr = df2.groupby(df2.key1)
gr
```

```python
gr.groups   # groups 속성 : 각 그룹 데이터의 인덱스를 저장한 속성
```

```python
gr.sum()
```

```python
df2.data1.groupby(df2.key1).sum()       # data1 열에 대해서만 그룹화
# df2.groupby('key1')['data1'].sum()  # 동일한 결과 but, 전체 데이터를 그룹화한 후 data1열에 대해서만 sum
```

```python
df2.groupby(df2.key1)[['data1', 'data2']].sum()
```

```python
df2.groupby(['key1', 'key2'])[['data1', 'data2']].sum() # .unstack('key2') # 인 경우 피봇테이블화
```

## **DataFrame Merge(합병, join)**
- ***merge*** 함수는 두 데이터 프레임의 공통 열 혹은 인덱스를 기준으로 두 개의 테이블을 합병
- 이 때 기준이 되는 열, 행의 데이터를 ***키(key)***라고 함

```python
mdf1 = pd.DataFrame({
    '고객번호': [1001, 1002, 1003, 1004, 1005, 1006, 1007],
    '이름': ['둘리', '도우너', '또치', '길동', '희동', '마이콜', '영희']
}, columns=['고객번호', '이름'])
mdf1
```

```python
mdf2 = pd.DataFrame({
    '고객번호': [1001, 1001, 1005, 1006, 1008, 1001],
    '금액': [10000, 20000, 15000, 5000, 100000, 30000]
}, columns=['고객번호', '금액'])
mdf2
```

- merge 함수로 위의 두 데이터프레임을 합치면 공통 열인 ***고객번호*** 열을 기준으로 데이터를 찾아서 합병함
- 이 때 기본적으로는 양쪽 데이터프레임에 모두 존재하는 키의 데이터만(교집합) 합병하는 ***inner join*** 방식을 사용(디폴트)

```python
pd.merge(mdf1, mdf2)
```

- ***outer join*** 방식은 키 값이 한쪽에만 있어도(합집합) 데이터를 합병함

```python
pd.merge(mdf1, mdf2, how='outer')
```

- ***left, right*** 방식은 각각 첫번째, 혹은 두번째 데이터프레임의 키 값에 대해서만 합병

```python
pd.merge(mdf1, mdf2, how='left')  # mdf1의 키 값에 대해서만 합병
# pd.merge(mdf1, mdf2, how='right')  # mdf2의 키 값에 대해서만 합병
```

- merge 명령어 대신 ***join*** 메소드를 사용할 수도 있음. 이 경우 행 인덱스의 값 기준으로 합병함.

```python
mdf1.join(mdf2, how='left', lsuffix='_L', rsuffix='_R')
```
