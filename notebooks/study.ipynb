# 기계학습의 시작
- 인간은 환경 내에서 일어나는 변화를 경험하면서 앞으로 발생하는 새로운 상황에 대한 예측이 가능
    - 귀가 시 자주 과일을 사가는 경우, 가격의 예상이 가능
    - 물고기를 잡아 판매하는 경우, 물고기의 길이를 통해 무게를 예상 가능
    - 부동산 관련업 종사자는, 평수를 통해 가격 유추가 가능

- 또한 환경 내에서 일어나는 상황을 통해 새로운 상황이 과거의 상황가 같은지 다른지 분류가 가능
    - 수많은 동물들을 보면서 산 사람은, 동물의 종류를 판별
    - 이진 분류
