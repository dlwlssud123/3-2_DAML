**빈칸 ①:** np.dot을 이용해 XW+b를 계산하세요.

**빈칸 ②:** 정답 T와 예측 Y의 차이를 제곱한 평균을 구하세요.

**목표 결과:** Y는 `[[2],[3],[4]]`, loss는 `0.0`
