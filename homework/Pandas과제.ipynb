```python
del df['Cabin']
df = df.dropna()
df = df.reset_index(drop=True)
df
```

## 3) 타이타닉 데이터 프레임 Human_classfi 컬럼 추가하기
- 남자이며, 19세 미만은 [Child_m]으로 설정
- 여자이며, 19세 미만은 [Child_f]으로 설정
- 남자이며, 19세 이상은 [Mr]으로 설정
- 여자이며, 19세 미만은 [Ms]으로 설정

```python
'''
    TODO:: Human_classfi 컬럼 추가 (위 조건에 맞는 값으로 넣을것)
    Hint : loc 함수 사용, 다중 조건 사용
'''

df['Human_classfi'] = 'Unknown'

df.loc[(df['Age'] < 19) & (df['Sex'] == 'male'), 'Human_classfi'] = 'Child_m'
df.loc[(df['Age'] < 19) & (df['Sex'] == 'female'), 'Human_classfi'] = 'Child_f'
df.loc[(df['Age'] >= 19) & (df['Sex'] == 'male'), 'Human_classfi'] = 'Mr'
df.loc[(df['Age'] >= 19) & (df['Sex'] == 'female'), 'Human_classfi'] = 'Ms'

df['Human_classfi'].value_counts()
```

## 4) 아래의 데이터 프레임을 타이타닉 데이터 프레임에 조인하기

```python
data_dict = {
    "Human_classfi" : ['Child_m', 'Child_f', 'Mr','Ms'],
    "HC_DES" : ['19세 미만의 남성', '19세 미만의 여성','19세 이상의 남성','19세 이상의 여성']
}
des_df = pd.DataFrame(data_dict)
des_df

```

```python
'''
    TODO:: 타이타닉 데이터 프레임에 위 데이터 프레임을 조인하기
    Hint : 키는 Human_classfi로 설정, merge, join 등 함수 사용하기
'''

df_merged = pd.merge(df, des_df, on = 'Human_classfi', how = 'inner')

df_merged.head(5)
```
