## **3. 인덱스 활용**

### **1) 1행 10열의 배열을 랜덤으로 5개 생성**

```python
arr1 = np.random.randint(1, 10, size = (1, 10))
arr2 = np.random.randint(1, 10, size = (1, 10))
arr3 = np.random.randint(1, 10, size = (1, 10))
arr4 = np.random.randint(1, 10, size = (1, 10))
arr5 = np.random.randint(1, 10, size = (1, 10))

arr1
```

```python
arr = np.zeros((10, 10))

for i in range(10):
    for j in range(10):
        if i == 0 or i == 9 or j == 0 or j == 9:
            arr[i][j] = 1
            
arr
```

### **2) 각 배열을 vstack을 이용하여 합치기**

```python
arr_stack = np.vstack((arr1, arr2, arr3, arr4, arr5))
arr_stack
```

### **3) 5행 10열의 배열을 랜덤으로 생성 후, 2)에서 합친 배열과 수평축으로 합치기**

```python
arr_random = np.random.randint(1, 10, size = (5, 10))
arr_hstack = np.hstack((arr_stack, arr_random))

arr_hstack
```

### **4) 3)을 전치**

```python
arr_hstack.T
```

### **2) 특정 열 인덱스를 기준으로 행을 정렬**

```python
arr_hstack_sorted = arr_hstack[arr_hstack[:, 2].argsort(), :]
arr_hstack_sorted

# 배열을 오름차순으로 정렬했을 때 인덱스 위치를 반환
# arr_hstack의 2번째 열을 기준으로 오름차순 정렬했을 때의 인덱스 위치를 반환
```

## **4. Random.Choice 활용**

### **1) 5행 6열 배열을 랜덤으로 생성**

- 각 행의 원소는 1~45사이의 랜덤 값이다.
- 각 행의 원소들은 행 별로 중복되지 않은 수로 구성한다.
- 1행 6열 배열을 5개 만들어 합치도록 코드를 작성하라.

```python
lotto_nums = np.random.choice( range(1, 47), size = (5, 6), replace = False)

lotto_nums
```

### **2) 1~45의 수를 랜덤으로 6개를 뽑는 코드를 작성**
- 단, 원하는 숫자 6개를 원하는 확률을 부여하여 생성할 것.

```python
arr = np.array(range(1, 47))
arr
```

```python
probs = np.zeros(46)
probs
```

```python
probs[3] = 0.17
probs[5] = 0.16
probs[10] = 0.18
probs[15] = 0.16
probs[20] = 0.17
probs[25] = 0.16
```

```python
probs
```

```python
np.random.choice(arr, size = 6, replace = False, p = probs)
```
